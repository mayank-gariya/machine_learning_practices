```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

# IterativeImputer is still experimental in scikit-learn, so we must enable it first
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

# 1. Start with a fresh reload of train data
train = pd.read_csv('/content/train.csv')

# 2. Separate target and features
id_col_data = train['id']
addicted_label_data = train['addicted_label']
train_features_only = train.drop(columns=['id', 'addicted_label'])

# Group the columns
mode_cols = ['gender', 'stress_level', 'academic_work_impact']
continuous_cols = [
    'social_media_hours', 'notifications_per_day', 'app_opens_per_day',
    'age', 'daily_screen_time_hours', 'gaming_hours',
    'sleep_hours', 'work_study_hours', 'weekend_screen_time'
]

# Ensure continuous columns are float types in the original feature set
for col in continuous_cols:
    train_features_only[col] = train_features_only[col].astype(float)

# 3. Perform train-test split *before* imputation
X_train, X_test, y_train, y_test = train_test_split(
    train_features_only, addicted_label_data, test_size=0.33, random_state=42, stratify=addicted_label_data
)

# Define the imputer transformer using IterativeImputer for continuous variables
imputer_transformer = ColumnTransformer(
    transformers=[
        ('mode_imputer', SimpleImputer(strategy='most_frequent'), mode_cols),
        ('mice_imputer', IterativeImputer(max_iter=10, random_state=42), continuous_cols)
    ],
    remainder='passthrough'
)

# Fit the imputer on X_train and transform both X_train and X_test
X_train_imputed_array = imputer_transformer.fit_transform(X_train)
X_test_imputed_array = imputer_transformer.transform(X_test)

# Reconstruct the DataFrames after imputation
processed_feature_names = mode_cols + continuous_cols

X_train = pd.DataFrame(
    X_train_imputed_array,
    columns=processed_feature_names,
    index=X_train.index
)
X_test = pd.DataFrame(
    X_test_imputed_array,
    columns=processed_feature_names,
    index=X_test.index
)

# Convert continuous columns back to float explicitly for both train and test
for col in continuous_cols:
    X_train[col] = X_train[col].astype(float)
    X_test[col] = X_test[col].astype(float)
```

```python
from sklearn.metrics import roc_auc_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
# plotting roc - auc
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt
```

```python
X_train['age'] = X_train['age'].astype(int)
X_test['age'] = X_test['age'].astype(int)
```

```python
# mapping each categorical values
gender_mapping = {'Male': 0, 'Female': 1}
stress_level_mapping = {'Low': 0, 'Medium': 1, 'High': 2}
academic_work_impact_mapping = {'Low': 0, 'Medium': 1, 'High': 2}

X_train['gender'] = X_train['gender'].map(gender_mapping)
X_train['stress_level'] = X_train['stress_level'].map(stress_level_mapping)
X_train['academic_work_impact'] = X_train['academic_work_impact'].map(academic_work_impact_mapping)

X_test['gender'] = X_test['gender'].map(gender_mapping)
X_test['stress_level'] = X_test['stress_level'].map(stress_level_mapping)
X_test['academic_work_impact'] = X_test['academic_work_impact'].map(academic_work_impact_mapping)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Initialize and train the RandomForestClassifier
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train,y_train)

# Make predictions on the test set
y_pred_rf = rf_model.predict(X_test)
y_pred_proba_rf = rf_model.predict_proba(X_test)[:, 1] # Probability of the positive class

# Evaluate the model
print("Random Forest Classifier Evaluation:\n")

# ROC AUC Score
roc_auc_rf = roc_auc_score(y_test, y_pred_proba_rf)
print(f"ROC AUC Score: {roc_auc_rf:.4f}")

# Confusion Matrix
conf_matrix_rf = confusion_matrix(y_test, y_pred_rf)
print("\nConfusion Matrix:")
print(conf_matrix_rf)

# Classification Report
class_report_rf = classification_report(y_test, y_pred_rf)
print("\nClassification Report:")
print(class_report_rf)

# Plotting ROC Curve
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_pred_proba_rf)
plt.figure(figsize=(8, 6))
plt.plot(fpr_rf, tpr_rf, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc_rf:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve - Random Forest')
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```

### Bagging Classifier Implementation and Evaluation

### LightGBM Classifier with 5-Fold Cross-Validation

```python
import lightgbm as lgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score, confusion_matrix, classification_report, roc_curve
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Identify categorical features (these were already defined
categorical_features = ['gender', 'stress_level', 'academic_work_impact']

# Initialize LightGBM Classifier
lgbm_model = lgb.LGBMClassifier(random_state=42, n_estimators=1000, learning_rate=0.05, num_leaves=31, n_jobs=-1)

# Set up Stratified K-Fold Cross-Validation
n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

roc_auc_scores = []
classification_reports = []
confusion_matrices = []
fpr_list = []
tpr_list = []

print(f"Starting {n_splits}-fold cross-validation for LightGBM...")

for fold, (train_index, val_index) in enumerate(skf.split(X_train, y_train)): # Use X_train and y_train from the previous cell
    print(f"\n--- Fold {fold+1}/{n_splits} ---")
    X_train_fold, X_val_fold = X_train.iloc[train_index].copy(), X_train.iloc[val_index].copy()
    y_train_fold, y_val_fold = y_train.iloc[train_index], y_train.iloc[val_index]

    # Convert categorical columns to 'category' dtype for LightGBM
    for col in categorical_features:
        X_train_fold[col] = X_train_fold[col].astype('category')
        X_val_fold[col] = X_val_fold[col].astype('category')

    # Train the LightGBM model
    lgbm_model.fit(X_train_fold, y_train_fold,
                    eval_set=[(X_val_fold, y_val_fold)],
                    eval_metric='auc',
                    callbacks=[lgb.early_stopping(stopping_rounds=50, verbose=False)], # Early stopping
                    categorical_feature=categorical_features)

    # Make predictions on the validation set
    y_pred_fold = lgbm_model.predict(X_val_fold)
    y_pred_proba_fold = lgbm_model.predict_proba(X_val_fold)[:, 1]

    # Evaluate current fold
    fold_roc_auc = roc_auc_score(y_val_fold, y_pred_proba_fold)
    roc_auc_scores.append(fold_roc_auc)

    fold_conf_matrix = confusion_matrix(y_val_fold, y_pred_fold)
    confusion_matrices.append(fold_conf_matrix)

    fold_class_report = classification_report(y_val_fold, y_pred_fold, output_dict=True)
    classification_reports.append(fold_class_report)

    # Store ROC curve for plotting (e.g., from the last fold, or average later)
    if fold == n_splits - 1: # Just taking the last fold's ROC for now
        fpr_lgbm, tpr_lgbm, _ = roc_curve(y_val_fold, y_pred_proba_fold)
        fpr_list.append(fpr_lgbm)
        tpr_list.append(tpr_lgbm)

print("\n--- Overall LightGBM Evaluation (5-Fold CV) ---")
print(f"Average ROC AUC: {np.mean(roc_auc_scores):.4f} (+/- {np.std(roc_auc_scores):.4f})")

# Display classification report and confusion matrix from the last fold for example
print("\nClassification Report (Last Fold):\n")
print(classification_reports[-1])

print("\nConfusion Matrix (Last Fold):\n")
print(confusion_matrices[-1])

# Plotting ROC Curve from the last fold
plt.figure(figsize=(8, 6))
plt.plot(fpr_list[-1], tpr_list[-1], color='blue', lw=2, label=f'ROC curve (area = {roc_auc_scores[-1]:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve - LightGBM (Last Fold)')
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```

```python
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score, confusion_matrix, classification_report, roc_curve
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Define feature groups
mode_cols = ['gender', 'stress_level', 'academic_work_impact']
continuous_cols = [
    'social_media_hours', 'notifications_per_day', 'app_opens_per_day',
    'age', 'daily_screen_time_hours', 'gaming_hours',
    'sleep_hours', 'work_study_hours', 'weekend_screen_time'
]
processed_feature_names = mode_cols + continuous_cols

# 2. Set up Stratified K-Fold Cross-Validation
n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

roc_auc_scores_xgb = []
confusion_matrices_xgb = []
classification_reports_xgb = []
fpr_list_xgb = []
tpr_list_xgb = []

print(f"Starting {n_splits}-fold cross-validation for XGBoost...")

# We use train_features_only and addicted_label_data to avoid pre-imputed leakage
for fold, (train_index, val_index) in enumerate(skf.split(train_features_only, addicted_label_data)):
    print(f"\n--- Fold {fold+ 1} / {n_splits} ---")

    # Split raw fold data
    X_train_fold = train_features_only.iloc[train_index].copy()
    X_val_fold = train_features_only.iloc[val_index].copy()
    y_train_fold = addicted_label_data.iloc[train_index]
    y_val_fold = addicted_label_data.iloc[val_index]

    # Ensure continuous columns are float types
    for col in continuous_cols:
        X_train_fold[col] = X_train_fold[col].astype(float)
        X_val_fold[col] = X_val_fold[col].astype(float)

    # 3. Define and fit Imputation Pipeline inside the fold
    imputer_transformer = ColumnTransformer(
        transformers=[
            ('mode_imputer', SimpleImputer(strategy='most_frequent'), mode_cols),
            ('mice_imputer', IterativeImputer(max_iter=10, random_state=42), continuous_cols)
        ],
        remainder='passthrough'
    )

    # Fit on training fold, transform both training and validation folds
    X_train_fold_imp = pd.DataFrame(imputer_transformer.fit_transform(X_train_fold), columns=processed_feature_names)
    X_val_fold_imp = pd.DataFrame(imputer_transformer.transform(X_val_fold), columns=processed_feature_names)

    # Convert continuous back to float, age to int
    for col in continuous_cols:
        X_train_fold_imp[col] = X_train_fold_imp[col].astype(float)
        X_val_fold_imp[col] = X_val_fold_imp[col].astype(float)
    X_train_fold_imp['age'] = X_train_fold_imp['age'].astype(int)
    X_val_fold_imp['age'] = X_val_fold_imp['age'].astype(int)

    # 4. Convert string/object columns to explicit 'category' dtype for XGBoost
    for col in mode_cols:
        X_train_fold_imp[col] = X_train_fold_imp[col].astype('category')
        X_val_fold_imp[col] = X_val_fold_imp[col].astype('category')

    # 5. Initialize XGBoost Classifier with modern early stopping parameter
    xgb_model = xgb.XGBClassifier(
        objective='binary:logistic',
        eval_metric='logloss',
        random_state=42,
        n_estimators=1000,
        learning_rate=0.05,
        n_jobs=-1,
        enable_categorical=True,
        early_stopping_rounds=50 # Placed here to prevent callback syntax issues
    )

    # Train the XGBoost model
    xgb_model.fit(
        X_train_fold_imp, y_train_fold,
        eval_set=[(X_val_fold_imp, y_val_fold)],
        verbose=False
    )

    # Predictions
    y_pred_fold_xgb = xgb_model.predict(X_val_fold_imp)
    y_pred_proba_fold_xgb = xgb_model.predict_proba(X_val_fold_imp)[:, 1]

    # Evaluation metrics
    fold_roc_auc_xgb = roc_auc_score(y_val_fold, y_pred_proba_fold_xgb)
    roc_auc_scores_xgb.append(fold_roc_auc_xgb)
    confusion_matrices_xgb.append(confusion_matrix(y_val_fold, y_pred_fold_xgb))
    classification_reports_xgb.append(classification_report(y_val_fold, y_pred_fold_xgb, output_dict=True))

    if fold == n_splits - 1:
        fpr_xgb, tpr_xgb, _ = roc_curve(y_val_fold, y_pred_proba_fold_xgb)
        fpr_list_xgb.append(fpr_xgb)
        tpr_list_xgb.append(tpr_xgb)

print("\n--- Overall XGBoost Evaluation (5-Fold CV) ---")
print(f"Average ROC AUC: {np.mean(roc_auc_scores_xgb):.4f} (+/- {np.std(roc_auc_scores_xgb):.4f})")
```

```python
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
import numpy as np
import pandas as pd

# 1. Load the raw test dataset
test = pd.read_csv('/content/test.csv')
test_ids = test['id']  # Save IDs for the final submission file
test_features_only = test.drop(columns=['id'])

# Define feature groups matching the training layout
mode_cols = ['gender', 'stress_level', 'academic_work_impact']
continuous_cols = [
    'social_media_hours', 'notifications_per_day', 'app_opens_per_day',
    'age', 'daily_screen_time_hours', 'gaming_hours',
    'sleep_hours', 'work_study_hours', 'weekend_screen_time'
]
processed_feature_names = mode_cols + continuous_cols

# Pre-convert raw test continuous fields to float
for col in continuous_cols:
    test_features_only[col] = test_features_only[col].astype(float)

# 2. Setup CV Loop & Placeholder array for test predictions
n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

# Array to accumulate out-of-fold test predictions: Shape (len(test), n_splits)
test_preds_folds = np.zeros((len(test_features_only), n_splits))

print(f"Starting {n_splits}-fold XGBoost training and test inference...")

for fold, (train_index, val_index) in enumerate(skf.split(train_features_only, addicted_label_data)):
    print(f"--- Processing Fold {fold + 1} / {n_splits} ---")

    # Split training & validation sets for this fold
    X_train_fold = train_features_only.iloc[train_index].copy()
    X_val_fold = train_features_only.iloc[val_index].copy()
    y_train_fold = addicted_label_data.iloc[train_index]
    y_val_fold = addicted_label_data.iloc[val_index]

    # Ensure continuous types
    for col in continuous_cols:
        X_train_fold[col] = X_train_fold[col].astype(float)
        X_val_fold[col] = X_val_fold[col].astype(float)

    # Define & Fit the Imputation Pipeline on THIS fold's training split only
    imputer_transformer = ColumnTransformer(
        transformers=[
            ('mode_imputer', SimpleImputer(strategy='most_frequent'), mode_cols),
            ('mice_imputer', IterativeImputer(max_iter=10, random_state=42), continuous_cols)
        ],
        remainder='passthrough'
    )

    # Transform train, validation, AND full test data using the fold's fitted imputer
    X_train_fold_imp = pd.DataFrame(imputer_transformer.fit_transform(X_train_fold), columns=processed_feature_names)
    X_val_fold_imp = pd.DataFrame(imputer_transformer.transform(X_val_fold), columns=processed_feature_names)

    # --- FIXED: The test features are now correctly imputed here ---
    X_test_fold_imp = pd.DataFrame(imputer_transformer.transform(test_features_only), columns=processed_feature_names)

    # Enforce precise data types post-imputation across all frames
    for col in continuous_cols:
        X_train_fold_imp[col] = X_train_fold_imp[col].astype(float)
        X_val_fold_imp[col] = X_val_fold_imp[col].astype(float)
        X_test_fold_imp[col] = X_test_fold_imp[col].astype(float)

    X_train_fold_imp['age'] = X_train_fold_imp['age'].astype(int)
    X_val_fold_imp['age'] = X_val_fold_imp['age'].astype(int)
    X_test_fold_imp['age'] = X_test_fold_imp['age'].astype(int)

    for col in mode_cols:
        X_train_fold_imp[col] = X_train_fold_imp[col].astype('category')
        X_val_fold_imp[col] = X_val_fold_imp[col].astype('category')
        X_test_fold_imp[col] = X_test_fold_imp[col].astype('category')

    # Initialize Model
    xgb_model = xgb.XGBClassifier(
        objective='binary:logistic',
        eval_metric='logloss',
        random_state=42,
        n_estimators=1000,
        learning_rate=0.05,
        n_jobs=-1,
        enable_categorical=True,
        early_stopping_rounds=50
    )

    # Train
    xgb_model.fit(
        X_train_fold_imp, y_train_fold,
        eval_set=[(X_val_fold_imp, y_val_fold)],
        verbose=False
    )

    # Predict probabilities for the test set using this fold's trained model
    # [:, 1] extracts the probability of the positive class ('addicted')
    test_preds_folds[:, fold] = xgb_model.predict_proba(X_test_fold_imp)[:, 1]

# 3. Average the test probabilities across all 5 folds
final_test_probabilities = test_preds_folds.mean(axis=1)

# 4. Generate the Kaggle submission DataFrame
submission = pd.DataFrame({
    'id': test_ids,
    'addicted_label': final_test_probabilities
})

# Save to CSV
submission.to_csv('submission.csv', index=False)
print("\n🎉 Submission file successfully created as 'submission.csv'!")
```
