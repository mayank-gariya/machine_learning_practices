```python
df.columns = df.columns.str.lower().str.strip()
```

```python
fault_columns = ['pastry', 'z_scratch', 'k_scatch', 'stains', 'dirtiness', 'bumps', 'other_faults']

def merge_fault_columns(row):
    for col in fault_columns:
        if row[col] == 1:
            return col
    return 'no_fault'

df['target_col'] = df.apply(merge_fault_columns, axis=1)
```

```python
df['target_col'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
import joblib
import os
```

```python
def engineer_data(df):
   '''
   this function engineer the data and create new features for better genralisation
   '''

   df['fault_width'] = df['x_maximum'] - df['x_minimum']
   df['fault_length'] = df['y_maximum'] - df['y_minimum']
   df['fault_area_estimate'] = df['fault_width'] * df['fault_length']
   df['thickness'] = df['steel_plate_thickness'] / (df['pixels_areas'] + 1e-5)

   return df
```

```python
def pipeline(df ,target_col='target_col',save_dir='model'):
   '''
   this function engineer the data and create new features for better genralisation
   '''
   # Create the save directory if it doesn't exist
   os.makedirs(save_dir, exist_ok=True)

   df = engineer_data(df)
   X = df.drop(columns=[target_col])
   y = df[target_col]

   #the 7 multi-class categorical targets into integers (0 to 6)
   label_encoder = LabelEncoder()
   y_encoded = label_encoder.fit_transform(y)
   joblib.dump(label_encoder, os.path.join(save_dir, 'label_encoder.joblib'))

   #Split into train and validation sets
   X_train, X_val, y_train, y_val = train_test_split(X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded)

   # Scale features (essential for Stacking Meta-models like Logistic Regression)
   scaler = StandardScaler()
   X_train_scaled = scaler.fit_transform(X_train)
   X_val_scaled = scaler.transform(X_val)
   joblib.dump(scaler, os.path.join(save_dir, 'scaler.joblib'))

   # Convert back to dataframes keeping track of engineered feature names
   feature_names = X.columns
   X_train_scaled = pd.DataFrame(X_train_scaled, columns=feature_names)
   X_val_scaled = pd.DataFrame(X_val_scaled, columns=feature_names)

   print(" Preprocessing complete. Scaler and Label Encoder saved.")
   return X_train_scaled, X_val_scaled, y_train, y_val
```

```python
from sklearn.ensemble import RandomForestClassifier, StackingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score
from xgboost import XGBClassifier

def train_ensemble_pipeline(data):
    # Get preprocessed data from our module
    X_train, X_val, y_train, y_val = pipeline(data)

    print("\nInitializing Base Models for the Ensemble...")

    # 1. Base Model 1: Random Forest (Bagging)
    rf_base = RandomForestClassifier(
        n_estimators=200,
        max_depth=10,
        random_state=42,
        n_jobs=-1
    )

    # 2. Base Model 2: XGBoost (Boosting)
    xgb_base = XGBClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=6,
        eval_metric='mlogloss',
        random_state=42,
        n_jobs=-1
    )

    # 3. The Meta-Learner (Stacking)
    # Stacking uses cross-validation predictions of base models to train the final estimator
    base_learners = [
        ('random_forest', rf_base),
        ('xgboost', xgb_base)
    ]

    meta_learner = LogisticRegression(max_iter=500)

    stacking_model = StackingClassifier(
        estimators=base_learners,
        final_estimator=meta_learner,
        cv=5, # 5-fold cross-validation for meta-training
        n_jobs=-1
    )

    # Train the end-to-end ensemble
    print("Training the Stacking Ensemble (this takes care of internal CV)...")
    stacking_model.fit(X_train, y_train)

    # Evaluate the model
    y_pred = stacking_model.predict(X_val)

    # Load label encoder to print actual class names instead of numbers (0-6)
    le = joblib.load('label_encoder.joblib')
    target_names = le.classes_

    print("\n=== ENSEMBLE PERFORMANCE ===")
    print(f"Validation Accuracy: {accuracy_score(y_val, y_pred):.4f}")
    print("\nClassification Report:")
    print(classification_report(y_val, y_pred, target_names=target_names))

    # Save the final ensemble model
    model_output_path = 'steel_faults_ensemble.joblib'
    joblib.dump(stacking_model, model_output_path)
    print(f"Final Stacking Ensemble model saved successfully to: {model_output_path}")
```

```python
import os
os.makedirs('model', exist_ok=True)
train_ensemble_pipeline(df)
```
