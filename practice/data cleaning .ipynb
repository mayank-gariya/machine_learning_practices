```python
import unicodedata

df.columns = (
    df.columns.map(lambda x: unicodedata.normalize('NFKD', x).encode('ascii', 'ignore').decode('utf-8'))
    .str.lower()
    .str.strip()
    .str.replace(r'[^a-z0-9_ ]', '', regex=True)
    .str.replace(r'\s+', '_', regex=True)
)
```

```python
df.rename(columns={'actualgross':'actual_gross', 'adjustedgross_in_2022_dollars':'gross_2012'},inplace=True)

df.head()
```

```python
df.isna().sum()
```

```python
df.drop(columns=['peak','all_time_peak'],inplace=True)
df
```

```python
for i in range(df.shape[0]):
  df['actual_gross'].values[i] = df['actual_gross'].str.strip().str.replace(r'[a-z ]','',regex=True).values[i][1:12]


df['actual_gross']
```

```python
for i in range(df.shape[0]):
  df['adjusted_gross_in_2022_dollars'].values[i] = df['adjusted_gross_in_2022_dollars'].str.strip().str.replace(r'[a-z ]','',regex=True).values[i][1:]


df['adjusted_gross_in_2022_dollars']
```

```python
for i in range(df.shape[0]):
  df['average_gross'].values[i] = df['average_gross'].str.strip().str.replace(r'[a-z ]','',regex=True).values[i][1:]


df['average_gross']
```

```python
df
```

```python
df['ref'] = df['ref'].str.extract(r'(\d+)').astype(float)

df['ref']
```

```python
df['tour_title'] = df['tour_title'].str.replace(r'[^a-zA-Z\s]', '', regex=True).str.strip()
df['tour_title']
```

```python
def parse_years(year_str):
    if isinstance(year_str, str):
        years = year_str.replace('–', '-').split('-')
        if len(years) == 1:
            start_year = int(years[0].strip())
            end_year = start_year
        else:
            start_year = int(years[0].strip())
            end_year = int(years[1].strip())
        return start_year, end_year
    return None, None

df[['start_year', 'end_year']] = df['years'].apply(lambda x: pd.Series(parse_years(x)))

df.drop(columns=['years'], inplace=True)
df[['start_year', 'end_year']].head()
```
