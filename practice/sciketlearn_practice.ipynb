```python
data = sns.load_dataset('titanic')
```

```python
data.drop(columns=['class',	'who'	,'adult_male'	,'deck',	'embark_town',	'alive',	'alone'],inplace=True)
```

```python
data
```

```python
X_train , X_test , y_train , y_test = train_test_split(data.drop(columns=['survived']),data['survived'],test_size=0.2)
```

```python
from sklearn.impute import SimpleImputer
```

```python
si_age = SimpleImputer()
si_embarked = SimpleImputer(strategy='most_frequent')

X_train_age = si_age.fit_transform(X_train[['age']])
X_train_embarked = si_embarked.fit_transform(X_train[['embarked']])

X_test_age = si_age.transform(X_test[['age']])
X_test_embarked = si_embarked.transform(X_test[['embarked']])
```

```python
ohe_sex = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
ohe_embarked = OneHotEncoder(sparse_output=False,handle_unknown='ignore')

X_train_sex = ohe_sex.fit_transform(X_train[['sex']])
X_train_embarked = ohe_embarked.fit_transform(X_train_embarked)

X_test_sex = ohe_sex.transform(X_test[['sex']])
X_test_embarked = ohe_embarked.transform(X_test_embarked)
```

```python
X_train_rem = X_train.drop(columns=['sex','age','embarked'])
X_test_rem = X_test.drop(columns=['sex','age','embarked'])
```

```python
X_transformed = np.concatenate((X_train_rem,X_train_age,X_train_embarked,X_train_sex),axis=1)
X_test_transformed = np.concatenate((X_test_rem,X_test_age,X_test_embarked,X_test_sex),axis=1)
```

```python
X_transformed.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf = DecisionTreeClassifier()
clf.fit(X_transformed,y_train)
```

```python
y_pred = clf.predict(X_test_transformed)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test,y_pred)
```

```python
from sklearn.datasets import make_regression
```

```python
# Create a simple dataset for linear regression with 1 feature
X_linear, y_linear = make_regression(n_samples=100, n_features=1, n_informative=1, noise=20, random_state=42)

# Plot the linear regression data
plt.scatter(X_linear, y_linear)
plt.xlabel("Feature")
plt.ylabel("Target")
plt.title("Data for Linear Regression")
plt.grid(True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test , y_train , y_test = train_test_split(X_linear,y_linear,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
beta = lr.coef_[0]
offset = lr.intercept_

print(f"Beta (Coefficient): {beta}")
print(f"Offset (Intercept): {offset}")
```

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
print('mse',mean_squared_error(y_test,y_pred))
print('r2',r2_score(y_test,y_pred))
```

```python
adjusted_r2 = 1 - (1-r2_score(y_test,y_pred))*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```

```python
X_linear, y_linear = make_regression(n_samples=100, n_features=2, n_informative=1, noise=20, random_state=42)
```

```python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

# Assuming X_linear and y_linear are already defined with n_features=2
# (from cell LeHNbKc5d0B1)

# Initialize and fit the Linear Regression model
lr_3d = LinearRegression()
lr_3d.fit(X_linear, y_linear)

# Prepare data for plotting the regression plane
x_surf, y_surf = np.meshgrid(np.linspace(X_linear[:, 0].min(), X_linear[:, 0].max(), 100),
                             np.linspace(X_linear[:, 1].min(), X_linear[:, 1].max(), 100))

z_surf = lr_3d.predict(np.array([x_surf.ravel(), y_surf.ravel()]).T).reshape(x_surf.shape)

# Create the 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the actual data points
ax.scatter(X_linear[:, 0], X_linear[:, 1], y_linear, c='blue', marker='o', label='Actual Data')

# Plot the regression plane
ax.plot_surface(x_surf, y_surf, z_surf, color='red', alpha=0.5, label='Regression Plane')

ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.set_zlabel('Target')
ax.set_title('3D Linear Regression Visualization')
ax.legend()
plt.show()
```
