```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
train = pd.read_csv('/content/Train.csv')
test = pd.read_csv('/content/Test.csv')
print('dataframe is loaded')
```

```python
train.shape , test.shape
```

```python
def get_null(test,train):
  test_null = test.isna().sum()
  train_null = test.isna().sum()

  print(test_null,train_null)

get_null(test,train)
```

```python
def get_dup(test,train):
  test_null = test.duplicated().sum()
  train_null = test.duplicated().sum()

  print(test_null,train_null)

get_dup(test,train)
```

```python
X = train.drop('Paddy_yield_kg',axis=1)
y = train['Paddy_yield_kg']
```

```python
# distribution of each column in tarin

columns = train.columns

for i , col in enumerate(columns):
   sns.distplot(train[col])
   plt.title(f'distribution of {col}')
   plt.show()
```

```python
# transforming using yeo jonson as a pipelie
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PowerTransformer

pipe = make_pipeline(
    PowerTransformer(method='yeo-johnson',standardize=True)
)
```

```python
X_transformed = pipe.transform(X)
X_transformed_df = pd.DataFrame(X_transformed, columns=X.columns)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_transformed, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score

model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)

print('mean squared error',mean_squared_error(y_test,y_pred))
print('r2 score',r2_score(y_test,y_pred))
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor()
model.fit(X_train,y_train)

Y_pred = model.predict(X_test)

print('mean squared error',mean_squared_error(y_test,y_pred))
print('r2 score',r2_score(y_test,y_pred))
```

```python
import lightgbm as lgb

# Initialize and train the LightGBM Regressor
lgbm_model = lgb.LGBMRegressor(random_state=42)
lgbm_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_lgbm = lgbm_model.predict(X_test)

# Evaluate the model
print('LightGBM - Mean Squared Error:', mean_squared_error(y_test, y_pred_lgbm))
print('LightGBM - R2 Score:', r2_score(y_test, y_pred_lgbm))
```

```python
import xgboost as xgb

# Initialize and train the XGBoost Regressor
xgb_model = xgb.XGBRegressor(random_state=42)
xgb_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_xgb = xgb_model.predict(X_test)

# Evaluate the model
print('XGBoost - Mean Squared Error:', mean_squared_error(y_test, y_pred_xgb))
print('XGBoost - R2 Score:', r2_score(y_test, y_pred_xgb))
```

```python
test_transformed = pipe.transform(test)

predictions = xgb_model.predict(test_transformed)

submission_df = pd.DataFrame({'Paddy_yield_kg': predictions})

submission_df.to_csv('submission.csv', index=False)

print('submission.csv created successfully!')
```
